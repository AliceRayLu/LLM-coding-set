# Tensor Parallel(TP)

> What is **Tensor Parallel**? 什么是张量并行？

In **Tensor Parallel (TP)**, a model layer is split across multiple GPUs, so each GPU stores and computes only a shard of the tensor.  
在 **张量并行（TP）** 中，一个模型层会被切分到多张 GPU 上，因此每张 GPU 只负责保存和计算其中一部分张量。

The **TP size** is the number of GPUs participating in this tensor split.  
**TP size** 表示参与这次张量切分的 GPU 数量。

A **rank** is the local index of a worker inside the TP group. In most training systems, each TP rank is mapped to **one GPU**.  
**rank** 是一个 worker 在 TP group 内部的本地编号。在大多数训练系统中，通常 **一个 TP rank 对应一张 GPU**。

For example, if **TP size = 4**, then the tensor is distributed across **4 GPUs**, and the TP ranks are usually **0, 1, 2, 3**.  
例如，如果 **TP size = 4**，那么张量会分布到 **4 张 GPU** 上，对应的 TP rank 通常就是 **0、1、2、3**。

So, when we discuss TP size restrictions, we are really asking: **how many GPUs are in the TP group, and how is the computation split across ranks?**  
所以，当我们讨论 TP size 的限制时，本质上是在问：**一个 TP group 里有多少张 GPU，以及计算是如何在这些 rank 之间切分的？**

不能随意指定`tp_size`。受到以下条件的制约：
